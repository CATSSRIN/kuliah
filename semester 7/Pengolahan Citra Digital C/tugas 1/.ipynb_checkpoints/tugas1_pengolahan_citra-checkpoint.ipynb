```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
import urllib.request
from skimage import data

os.makedirs('input', exist_ok=True)
os.makedirs('output', exist_ok=True)

if not os.path.exists('input/selfie.jpg'):
    cv2.imwrite('input/selfie.jpg', data.camera())

if not os.path.exists('input/rice.png'):
    try:
        cv2.imwrite('input/rice.png', data.rice())
    except:
        cv2.imwrite('input/rice.png', data.coins())

if not os.path.exists('input/peppers.png'):
    try:
        urllib.request.urlretrieve('https://raw.githubusercontent.com/opencv/opencv/master/samples/data/fruits.jpg', 'input/peppers.png')
    except:
        cv2.imwrite('input/peppers.png', cv2.cvtColor(data.astronaut(), cv2.COLOR_RGB2BGR))
```

```python
I = cv2.imread('input/selfie.jpg', cv2.IMREAD_GRAYSCALE)
T = 128
BW = (I >= T).astype(np.uint8) * 255
cv2.imwrite('output/1_thresholding.png', BW)

plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.imshow(I, cmap='gray')
plt.title('Grayscale')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(BW, cmap='gray')
plt.title('Thresholding')
plt.axis('off')
plt.show()
```

```python
I = cv2.imread('input/selfie.jpg', cv2.IMREAD_GRAYSCALE)
negative = 255 - I
cv2.imwrite('output/2_negative.png', negative)

plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.imshow(I, cmap='gray')
plt.title('Original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(negative, cmap='gray')
plt.title('Negative')
plt.axis('off')
plt.show()
```

```python
I = cv2.imread('input/selfie.jpg', cv2.IMREAD_GRAYSCALE)
b = 50
J = np.clip(I.astype(np.float64) + b, 0, 255).astype(np.uint8)
cv2.imwrite('output/3_brightening.png', J)

plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.imshow(I, cmap='gray')
plt.title('Original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(J, cmap='gray')
plt.title('Brightening')
plt.axis('off')
plt.show()
```

```python
I = cv2.imread('input/peppers.png')
gray = cv2.cvtColor(I, cv2.COLOR_BGR2GRAY)
cv2.imwrite('output/4_grayscale.png', gray)

plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(I, cv2.COLOR_BGR2RGB))
plt.title('RGB')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(gray, cmap='gray')
plt.title('Grayscale')
plt.axis('off')
plt.show()
```

```python
A = cv2.imread('input/selfie.jpg', cv2.IMREAD_GRAYSCALE).astype(np.float64) / 255.0
B = cv2.imread('input/rice.png', cv2.IMREAD_GRAYSCALE)
B = cv2.resize(B, (A.shape[1], A.shape[0])).astype(np.float64) / 255.0
C = (A + B) / 2.0
cv2.imwrite('output/5_penjumlahan.png', (C * 255).astype(np.uint8))

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(A, cmap='gray')
plt.title('Citra A')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(B, cmap='gray')
plt.title('Citra B')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(C, cmap='gray')
plt.title('Average')
plt.axis('off')
plt.show()
```

```python
A = cv2.imread('input/selfie.jpg', cv2.IMREAD_GRAYSCALE)
B = A.copy()
B[80:150, 80:150] = 0
D = cv2.absdiff(A, B)
cv2.imwrite('output/6_difference.png', D)

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(A, cmap='gray')
plt.title('Citra A')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(B, cmap='gray')
plt.title('Citra B')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(D, cmap='gray')
plt.title('Difference')
plt.axis('off')
plt.show()
```

```python
A = cv2.imread('input/selfie.jpg', cv2.IMREAD_GRAYSCALE) > 128
B = np.zeros(A.shape, dtype=bool)
B[80:180, 80:180] = True
AND_result = A & B
OR_result = A | B
NOT_result = ~A

cv2.imwrite('output/7_boolean_and.png', (AND_result.astype(np.uint8) * 255))
cv2.imwrite('output/7_boolean_or.png', (OR_result.astype(np.uint8) * 255))
cv2.imwrite('output/7_boolean_not.png', (NOT_result.astype(np.uint8) * 255))

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(AND_result, cmap='gray')
plt.title('AND')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(OR_result, cmap='gray')
plt.title('OR')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(NOT_result, cmap='gray')
plt.title('NOT')
plt.axis('off')
plt.show()
```

```python
I = cv2.imread('input/selfie.jpg'cv2.IMREAD_GRAYSCALE)
tx = 50
ty = 30
h, w = I.shape
M = np.float32([[1, 0, tx], [0, 1, ty]])
J = cv2.warpAffine(I, M, (w, h))
cv2.imwrite('output/8_translasi.png', J)

plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.imshow(I, cmap='gray')
plt.title('Original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(J, cmap='gray')
plt.title('Translation')
plt.axis('off')
plt.show()
```

```python
I = cv2.imread('input/selfie.jpg', cv2.IMREAD_GRAYSCALE)
h, w = I.shape
center = (w / 2, h / 2)
M1 = cv2.getRotationMatrix2D(center, 45, 1.0)
J1 = cv2.warpAffine(I, M1, (w, h))
M2 = cv2.getRotationMatrix2D(center, 90, 1.0)
J2 = cv2.warpAffine(I, M2, (w, h))

cv2.imwrite('output/9_rotasi_45.png', J1)
cv2.imwrite('output/9_rotasi_90.png', J2)

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(I, cmap='gray')
plt.title('Original')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(J1, cmap='gray')
plt.title('45°')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(J2, cmap='gray')
plt.title('90°')
plt.axis('off')
plt.show()
```

```python
I = cv2.imread('input/selfie.jpg', cv2.IMREAD_GRAYSCALE)
horizontal = np.fliplr(I)
vertical = np.flipud(I)

cv2.imwrite('output/10_flipping_horizontal.png', horizontal)
cv2.imwrite('output/10_flipping_vertical.png', vertical)

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(I, cmap='gray')
plt.title('Original')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(horizontal, cmap='gray')
plt.title('Horizontal Flip')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(vertical, cmap='gray')
plt.title('Vertical Flip')
plt.axis('off')
plt.show()
```

```python
I = cv2.imread('input/selfie.jpg', cv2.IMREAD_GRAYSCALE)
besar = cv2.resize(I, (0, 0), fx=2, fy=2)
kecil = cv2.resize(I, (0, 0), fx=0.5, fy=0.5)

cv2.imwrite('output/11_scaling_2x.png', besar)
cv2.imwrite('output/11_scaling_0.5x.png', kecil)

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(I, cmap='gray')
plt.title('Original')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(besar, cmap='gray')
plt.title('2×')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(kecil, cmap='gray')
plt.title('0.5×')
plt.axis('off')
plt.show()
```
