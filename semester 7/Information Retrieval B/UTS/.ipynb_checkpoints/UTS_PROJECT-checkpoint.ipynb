```python
# ACA DATASET
df = pd.read_csv('google_books_dataset.csv')

print(f"Total baris awal: {len(df)}")

# FILTERING KOLOM
df = df[['title', 'authors', 'categories', 'description']].copy()

# PEMBERSIHAN DATA (MENGHAPUS NULL)
df = df.dropna(subset=['title', 'authors', 'categories', 'description'])

df = df.reset_index(drop=True)

print(f"Total baris setelah dibersihkan (siap diproses): {len(df)}")

# PENGGABUNGAN TEKS (TEXT PREPROCESSING)
df['konten'] = (
    df['title'].astype(str) + " " +
    df['authors'].astype(str) + " " +
    df['categories'].astype(str) + " " +
    df['description'].astype(str)
)

df.head()
```

```python
# Case Folding & Cleaning
def cleaning_teks(text):
    text = str(text).lower()                               # Case folding
    text = re.sub(r'<.*?>', ' ', text)                     # Hapus tag HTML (<br /> dll)
    text = re.sub(r'http\S+|www\S+|https\S+', '', text)    # Hapus link URL
    text = re.sub(r'\d+', '', text)                        # Hapus angka
    text = re.sub(r'[^\w\s]', '', text)                    # Hapus tanda baca
    text = re.sub(r'\s+', ' ', text).strip()               # Rapikan spasi berlebih
    return text

# Mengaplikasikan fungsi cleaning
df['tahap1_konten_cleaning'] = df['konten'].apply(cleaning_teks)

# Membandingkan teks sebelum dan sesudah di-cleaning
df[['konten', 'tahap1_konten_cleaning']].head()
```

```python
# Tokenisasi sederhana memisahkan berdasarkan spasi
# Menggunakan 'konten_bersih' yang berasal dari cell di atasnya
df['tahap2_token'] = df['tahap1_konten_cleaning'].apply(lambda x: x.split())

# Menampilkan bentuk array/list dari tokenisasi
df[['tahap1_konten_cleaning', 'tahap2_token']].head()
```

```python
#Stopword Removal
stop_words_eng = set(stopwords.words('english'))

def remove_stopwords(tokens):
    return [kata for kata in tokens if kata not in stop_words_eng]

df['tahap3_stopword'] = df['tahap2_token'].apply(remove_stopwords)

# Menampilkan perbandingan token sebelum dan sesudah stopword dibuang
df[['tahap2_token', 'tahap3_stopword']].head()
```

```python
# Normalisasi (Lemmatization)
# Inisialisasi Lemmatizer (berbasis kamus)
lemmatizer = WordNetLemmatizer()

def apply_lemmatization(tokens):
    # Mengecek setiap kata ke dalam kamus WordNet untuk mengembalikannya ke kata dasar yang VALID
    kata_lemma = [lemmatizer.lemmatize(kata) for kata in tokens]
    return ' '.join(kata_lemma)

# Mengaplikasikan ke dataset
df['tahap4_final'] = df['tahap3_stopword'].apply(apply_lemmatization)

# Lihat perbedaannya! Kata yang dihasilkan akan jauh lebih enak dibaca manusia
# dibanding hasil Stemming yang terpotong-potong.
df[['tahap3_stopword', 'tahap4_final']].head()
```

```python
#BACK UP DATA
print("Menyiapkan data teks final untuk backup...")

# Kita hanya mem-backup kolom identitas buku dan kolom teks yang sudah dinormalisasi
df_backup = df[['title', 'authors', 'categories', 'tahap4_final']].copy()

# Menyimpan ke CSV
nama_file = 'backup_buku_dinormalisasi.csv'
df_backup.to_csv(nama_file, index=False)

print(f"✅ Selesai! File '{nama_file}' berhasil disimpan.")
```

```python
# TF - IDF
print("Memulai pembobotan TF-IDF...")

# Parameter dibiarkan kosong karena stopword sudah dibuang manual
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(df['tahap4_final'])

print("TF-IDF Vectorizer berhasil dibentuk!")
print(f"Dimensi Matriks: {tfidf_matrix.shape}")
```

```python
# COSINE SIMILARITY


print("Menghitung matriks kemiripan Cosine Similarity...")
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)

print(f"Dimensi Cosine Similarity: {cosine_sim.shape}")
print("✅ Matriks siap digunakan untuk mencari rekomendasi!")
```

```python
# ==========================================
# FUNGSI PENCARIAN & REKOMENDASI BUKU
# ==========================================
def singkat_judul(judul, panjang=50):
    judul = str(judul)
    return judul[:panjang] + "..." if len(judul) > panjang else judul

def rekomendasi_buku(keyword, kategori=None):
    # 1. Filter berdasarkan kategori jika ada
    if kategori:
        df_filter = df[df['categories'].astype(str).str.contains(kategori, case=False, na=False)]
        if df_filter.empty:
            print("❌ Tidak ada buku pada kategori tersebut.")
            return
    else:
        df_filter = df

    # 2. Mencari buku acuan berdasarkan kata kunci di kolom judul
    kandidat = df_filter[df_filter['title'].astype(str).str.contains(keyword, case=False, na=False)]
    if kandidat.empty:
        print("❌ Tidak ada buku dengan kata kunci tersebut.")
        return

    # 3. Mengambil buku pertama yang cocok
    idx = kandidat.index[0]
    pos = df.index.get_loc(idx)

    print(f"\n📘 BUKU ACUAN:\n-> {df.loc[idx, 'title']} - {df.loc[idx, 'authors']}")

    # 4. Mengambil skor kemiripan dari model Cosine Similarity
    sim_scores = list(enumerate(cosine_sim[pos]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    rekom_indices = [i for i, score in sim_scores if i != pos]

    # 5. Memastikan rekomendasi tetap berada dalam kategori yang difilter
    if kategori:
        rekom_indices = [i for i in rekom_indices if i in df_filter.index]

    # 6. Mengambil Top-5 rekomendasi
    rekom = df.loc[rekom_indices[:5], ['title', 'authors', 'categories']]

    print("\n📚 TOP 5 REKOMENDASI BUKU SERUPA:")
    for i, row in enumerate(rekom.itertuples(), 1):
        print(f"{i}. {singkat_judul(row.title)} - {row.authors} | {row.categories}")


# ==========================================
# TEST RUN SISTEM (INTERAKTIF)
# ==========================================
print("\n===== PENGUJIAN SISTEM =====")

# Meminta input langsung dari pengguna
input_keyword = input("Masukkan kata kunci judul buku : ")
input_kategori = input("Masukkan kategori (tekan Enter untuk mengosongkan) : ")

# Menangani jika pengguna hanya menekan Enter pada kategori
input_kategori = input_kategori if input_kategori.strip() else None

# Menjalankan fungsi dengan data yang diketik pengguna
rekomendasi_buku(input_keyword, input_kategori)
```
